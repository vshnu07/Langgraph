```python
from langgraph.graph import StateGraph,START,END
from langchain_ollama import ChatOllama
from typing import TypedDict
```

```python
model=ChatOllama(model="llama3.2:latest")
```

```python
class llm(TypedDict):
    topic:str
    outline:str
    blog:str
```

```python
def llm_outline(state:llm)->llm:
    topic=state['topic']
    prompt=f'Generate a detail outling blog on a topic-{topic}'
    outline=model.invoke(prompt).content
    state["outline"]=outline

    return state
```

```python
def llm_blog(state:llm)->llm:
    topic=state['topic']
    outline=state["outline"]
    prompt=f'write a detail blog on a topic - {topic} using the following outline \n{outline}'
    blog=model.invoke(prompt).content
    state['blog']=blog

    return state
```

```python
graph=StateGraph(llm)
```

```python
graph.add_node("llm_outline",llm_outline)
graph.add_node("llm_blog",llm_blog)

graph.add_edge(START,'llm_outline')
graph.add_edge('llm_outline','llm_blog')
graph.add_edge('llm_blog',END)

workflow=graph.compile()
```

```python
intial_state={"topic":"Rise of AI in India"}
final_state=workflow.invoke(intial_state)
print(final_state)
```
