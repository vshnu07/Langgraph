```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class BMI(TypedDict):
    height_m:float
    weight_kg:float
    bmi:float
    cattegory:str
```

```python
def calculate_bmi(state:BMI)->BMI:
    height = state['height_m']
    weight=state['weight_kg']
    bmi=weight/(height**2)
    state['bmi']=round(bmi,2)

    return state
```

```python
def lable_bmi(state:BMI)->BMI:
    bmi=state['bmi']

    if(bmi<18.5):
        state['cattegory']="Underweight"
    elif(18.5<=bmi<25):
        state['cattegory']="Normal"
    elif(25<=bmi<30):
        state['cattegory']="Overweight"
    else:
        state['cattegory']="Obese"
    return state
```

```python
graph=StateGraph(BMI)


graph.add_node('calculate_bmi',calculate_bmi)
graph.add_node('lable_bmi',lable_bmi)

graph.add_edge(START,'calculate_bmi')
graph.add_edge('calculate_bmi','lable_bmi')
graph.add_edge('lable_bmi',END)


workflow=graph.compile()
```

```python
workflow.invoke({'height_m':1.8,'weight_kg':80})
```
